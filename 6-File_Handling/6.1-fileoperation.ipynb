File Operation- Read and write Files

Python provides built-in functions and methods to read from and write to files, both text and binary files.

```python
## Read the whole file content

with open('example.txt', 'r') as file:
    content = file.read()
    for row in content:
        print(row)
```

```python
## Read the file from line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
## Writing the File(Overwriting)

with open('example.txt', 'w') as file:
    file.write('Hello World\n')
    file.write('This is the new line\n')
```

```python
## Writing the file (Without overwriting)

with open('example.txt', 'a') as file:
    file.write('Append Operation taking place\n')
    file.write('Without overriding the previous data of the file\n')
```

```python
## Write a list of lines

lines = ['First line \n', 'Second line \n', 'Third line \n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
## Binary files

# writing to the binary file
data = b'\x00\x01\x02\x03\x04'

with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# reading the binary file

with open('example.bin', 'rb') as binFile:
    content = binFile.read()
    print(content)
```

```python
# Read the content from the source file and store it on destination file

with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
# Read a text file and count the number of lines, words and characters

def count_text_file(file_path):
    word_count = 0
    with open(file_path, 'r') as file:
        lines = file.readlines()
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
        return len(lines), word_count, char_count




file_path = 'example.txt'
lines, words, chars = count_text_file(file_path)

print(f'Lines: {lines}, Words: {words}, chars: {chars}')
```
