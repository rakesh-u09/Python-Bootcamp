# File handling

```python
import os
new_directory = "package"

os.mkdir(new_directory)
print(f'new Directry {new_directory} created')
```

```python
# Listing the files and directry
items = os.listdir('.')
print(items)
```

```python
# checking if a path is directry or file

path = 'example.txt'
if os.path.isfile(path):
    print(f'The path {path} is a file')
elif os.path.isdir(path):
    print(f'The path {path} is a directry')
else:
    print(f'The path is neither a file nor a directry')
```

```python
# getting the absolute path

relative_path = 'example.txt'

absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
