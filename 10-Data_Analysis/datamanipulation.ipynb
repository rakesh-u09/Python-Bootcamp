Data Manipulation and Analysis with Pandas

Data manipulation and analysis are the tasks any data science or data analysis project.


```python
import pandas as pd
```

```python
reviews = pd.read_csv('wine_data.csv', index_col=0)
reviews.head()

reviews.isnull().sum()
```

```python
# fill in missing values with the mean of the column
reviews['price'] = reviews['price'].fillna(reviews['price'].mean())
```

```python
reviews.dtypes
```

```python
# renaming the columns

reviews.rename(columns={'points': 'score', 'price': 'cost'}, inplace=True)
reviews.head()

reviews.tail()
```

```python
# change data type of a column
reviews['score_Points'] = reviews.score.astype(float)

reviews.head()
```

```python
# apply function

reviews['New Value'] = reviews.score.apply(lambda x: x*2)

reviews.head()
```

```python
# Data aggregating and grouping

grouped_mean = reviews.groupby('country')['score'].mean()

print(grouped_mean)
```

```python
grouped_sum = reviews.groupby(['country','region_1'])['score_Points'].mean()

print(grouped_sum)
```

```python
# aggregate multiple function

mul_agreegator = reviews.groupby('province')['cost'].agg(['mean', 'sum', 'count', 'max'])

print(mul_agreegator)
```
