Built-in Numpy methods are 100-1000 times faster than python implementation

```python
import numpy as np
import time
```

```python
# test ans compare  performance of numpy vs plain python loop - dot product

n = 10000000

a = [1] * n
b = [2] * n

start = time.time()
c = [0] * n

for i in range(n):
    c[i] = a[i] * b[i]
end = time.time()
python_list_time = end - start
print("Python list took: ", python_list_time)

numpy_a = np.ones(n)
numpy_b = np.ones(n) * 2
start = time.time()
numpy_c = np.dot(numpy_a, numpy_b)
numpy_time = time.time() - start
print("Numpy took: ", numpy_time)

print("Speedup: ", python_list_time / numpy_time)
```

```python
#   create 2 matrices using numpy
A = np.array([[1, 2, 3],[0, 1, 0],[0, 0, 1]])
B = np.array([[0, 1, 0],[1, 0, 0],[1, 2, 1]])
```

```python
# find the sum of two matrix
A + B
```

```python
# difference of two matrix
A - B
```

```python
# finding the product of two matrix
A@B
```

```python
# Finding transpose of a matrix
A.T
print(np.transpose(A))
```

```python
# Finding determinant of the matrix
print(np.linalg.det(A))
```

```python
# finding the inverse of a matrix
print(np.linalg.inv(A))
```

```python
# Finding the trace of a matrix
print(np.linalg.trace(A))
```

```python
# Finding rank of a matrix
print(np.linalg.matrix_rank(A))
```

```python
# FInding eigen value and Eigen vector of a matrix
print(np.linalg.eig(A))
```
