```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Create a simple line plot
plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Show the plot
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# customize the plot
plt.plot(x, y, color='green', marker='o', linestyle='--', linewidth=2, markersize=8)
plt.title("Customized Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Show the plot
plt.show()
```

```python
# multiple lines on the same plot
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]
y3 = [3, 5, 7, 9, 12]

plt.plot(x, y1, label='Line 1', color='blue', marker='o')
plt.plot(x, y2, label='Line 2', color='red', marker='s')
plt.plot(x, y3, label='Line 3', color='green', marker='^')
plt.title("Multiple Lines on the Same Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()  # Show legend
# Show the plot
plt.show()
```

```python
# Bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 6]
plt.bar(categories, values, color='grey', edgecolor='black')
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
# Show the plot
plt.show()
```

Histogram

Histogram are used to represent the distribution of a dataset, They divide the data into bins and count the number of data points.

```python
x = [1, 1, 2, 3, 3, 3, 4, 5, 5]

plt.hist(x, bins=5, color='purple', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
# Show the plot
plt.show()
```

```python
# scatter plot
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.scatter(x, y, color='orange', marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Show the plot
plt.show()
```
