Pandas DataFrame and Series

Pandas is a powerful data manipulation library in python.

```python
import pandas as pd
```

```python
# DataFrame

'''A DataFrame is a table. It contains an array of individual entries, 
each of which has a certain value. Each entry corresponds to a row (or record) and a column.'''

print(pd.DataFrame([[1, 2], [3, 4]], columns=['A', 'B'], index=['row1', 'row2']))
```

```python
df = pd.DataFrame({'Bob':['I like it', 'It was good'],
                   'Sue':['I did not like it', 'It was not good']}, index=['Product_A', 'Product_B'])

print(df)
```

```python
# Series

print(pd.Series([1,2,3,4,5,6,7,8], dtype='Int32', index=['a','b','c','d','e','f','g','h']))
```
