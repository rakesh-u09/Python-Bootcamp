Abstraction

is the concept of hiding the complex implementation details and showing only the necessary feature of the object.

```python
from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass # Abstact method

class English(Greet):
    
    def say_hello(self):
        print("Hello from English")


# creating the object

obj1 = English()
obj1.say_hello()
```
