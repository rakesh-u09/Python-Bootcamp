Encapsulation

is the concept of wrapping data(variable) and methods(functions) together as a single unit.

```python
class Person:
    def __init__(self, name, age):
        self.name = name   ## public property
        self.__age = age
    
person = Person("Rakesh", 25)
print(person.name)

print(person.age)  # can't access private property outside the class
```

```python
class Person:
    def __init__(self, name, age):
        self._name = name   ## protected property
        self.__age = age    ## private property
    
person = Person("Rakesh", 25)
print(person.name)
```

```python
## Encapsulation with getter and setter methods

class Car:
    def __init__(self, wheels, doors):
        self.__wheels = wheels # private variable
        self.__doors = doors

    ## getter method for wheels
    def get_wheels(self):
        return self.__wheels
    
    ## getter method for doors
    def get_doors(self):
        return self.__doors
    
    def set_wheels(self, wheels):
        self.__wheels = wheels
    
    def set_doors(self, doors):
        self.__doors = doors


## creating the object

car1 = Car(4, 4)
print(car1.get_wheels())

dir(car1)
```
