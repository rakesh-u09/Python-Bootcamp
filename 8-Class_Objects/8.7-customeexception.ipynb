Custome ecxeption (Raise and Throw an exception)

```python
class DobException(Exception):
    pass
```

```python
try:
    age = int(input("Enter the your age:"))

    if (age <= 30 and age >= 20):
        print(f'your age is {age} and you can apply for the job')
    else:
        raise DobException
except DobException:
    print("you can't apply for this")
```
