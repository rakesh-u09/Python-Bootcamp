Inheritance in python

Inheritance that allows you to inherit attributes and methods from another class.

```python
## Base class

class Car:
    def __init__(self, doors, windows, enginetype):
        self.doors = doors
        self.windows = windows
        self.enginetype = enginetype

    def apply_brake(self):
        print('brake applied')
    
    def apply_accelaration(self):
        print('accelarion applied')


class Tesla(Car):
    def __init__(self, doors, windows, enginetype, isselfdriving):
        super().__init__(doors, windows, enginetype)
        self.is_self_driving = isselfdriving
    def switch_self_driving_mode(self):
        self.is_self_driving = True
        print('self driving mode activated')

    def switch_off_self_driving_mode(self):
        self.is_self_driving = False

## creating the objects

xModel = Tesla(3, 4, 'ev', True)

xModel.apply_brake()
xModel.switch_self_driving_mode()
```
