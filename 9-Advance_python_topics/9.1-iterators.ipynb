Iterators

an iterator is an object that allows you to traverse a data stream or collection one element at a time

__iter__(): Returns the iterator object itself. This allows iterators to be used where an iterable is expected (like in a for loop).

__next__(): Returns the next value from the sequence. If there are no more items left, it must raise a StopIteration exception to signal the end of the

```python
## iterator

mylist = ["apple", "mango", "banana", "orange"]

iterator = iter(mylist)

next(iterator)
```

```python
class CountUpTo:
    """An iterator that counts up to a specified maximum limit."""
    def __init__(self, max_limit):
        self.max_limit = max_limit
        self.current = 1

    def __iter__(self):
        return self  # An iterator must return itself

    def __next__(self):
        if self.current <= self.max_limit:
            result = self.current
            self.current += 1
            return result
        else:
            raise StopIteration  # Signals completion

counter = CountUpTo(3)
for num in counter:
    print(num)
```
